### Scalable & Production-Ready Pattern (Loop-Based)

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
base_path = "abfss://bronze@swap01storageaccount.dfs.core.windows.net/"

folders = [
    "AdventureWorks_Calendar",
    "AdventureWorks_Customers",
    "AdventureWorks_Product_Categories",
    "AdventureWorks_Product_Subcategories",
    "AdventureWorks_Products",
    "AdventureWorks_Returns",
    "AdventureWorks_Sales_2015",
    "AdventureWorks_Sales_2016",
    "AdventureWorks_Sales_2017",
    "AdventureWorks_Territories"
]

dfs = {}

for folder in folders:
    dfs[folder] = spark.read.format("csv") \
        .option("header", True) \
        .option("inferSchema", True) \
        .load(base_path + folder)
```

```python
[d for d in globals() if d.startswith("df_")]
```

```python
for name, df in dfs.items():
    print(f"Displaying {name}")
    df.display()
```

```python
dfs["AdventureWorks_Calendar"].display()
```

```python
dfs["AdventureWorks_Calendar"] = dfs["AdventureWorks_Calendar"].withColumn('Month', month(col('Date')))
dfs["AdventureWorks_Calendar"].display()
```
